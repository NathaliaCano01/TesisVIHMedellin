```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configuración de estilo profesional
sns.set_theme(style="white")  # Fondo blanco minimalista
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'
plt.rcParams['grid.alpha'] = 0.3

# Datos
modelos = ['XGBoost', 'Random Forest', 'Holt-Winters']
metricas = ['RMSE', 'MAE']
valores = np.array([
    [0.08, 0.07],      # XGBoost
    [17.25, 13.42],     # Random Forest
    [21.51, 16.99]      # Holt-Winters
])

# Crear figura
fig, ax = plt.subplots(figsize=(10, 6), dpi=300)

# Colores profesionales
colores = ['#1a6fdf', '#ff7f0e']  # Azul corporativo y naranja

# Espaciado entre grupos
espacio_barras = 0.4
altura_barra = 0.35
y_pos = np.arange(len(modelos)) * (len(metricas) * altura_barra * 1.5)

# Dibujar barras horizontales
for i, modelo in enumerate(modelos):
    for j, metrica in enumerate(metricas):
        ax.barh(
            y_pos[i] + j * altura_barra,
            valores[i, j],
            height=altura_barra,
            color=colores[j],
            edgecolor='black',
            linewidth=0.8,
            alpha=0.9,
            label=metrica if i == 0 else ""
        )
        # Añadir etiquetas de valor
        ax.text(
            valores[i, j] + 0.5,
            y_pos[i] + j * altura_barra,
            f'{valores[i, j]:.2f}',
            va='center',
            ha='left',
            fontsize=10,
            fontweight='bold'
        )

# Personalización avanzada
ax.set_title('Comparación de Métricas de Error por Modelo\n(Validación Cruzada)',
            fontsize=14, fontweight='bold', pad=20, color='#333333')
ax.set_xlabel('Valor del Error', fontsize=12, labelpad=10, color='#333333')
ax.set_ylabel('Modelos', fontsize=12, labelpad=10, color='#333333')

# Ejes y ticks
ax.set_yticks(y_pos + altura_barra/2)
ax.set_yticklabels(modelos, fontsize=11, color='#333333')
ax.set_xticks(np.arange(0, 25, 5))  # Ajustar según tus datos
ax.xaxis.grid(True, linestyle='--', alpha=0.4)
ax.yaxis.grid(False)

# Eliminar bordes no esenciales
for spine in ['top', 'right']:
    ax.spines[spine].set_visible(False)

# Leyenda mejorada
handles, labels = ax.get_legend_handles_labels()
ax.legend(handles[:2], metricas,
         frameon=True,
         framealpha=0.9,
         facecolor='white',
         edgecolor='#eeeeee',
         fontsize=11,
         loc='lower right',
         bbox_to_anchor=(1, 1))

# Ajustar márgenes
plt.tight_layout()

# Guardar en alta calidad (opcional)
plt.savefig('comparacion_horizontal_profesional.png', dpi=300, bbox_inches='tight')

plt.show()
```
